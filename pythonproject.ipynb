```python
def load_data(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
    return data
```

```python
## Write a function to display users and their connections
def display_users(data):
  user_dict = {user['id']: user for user in data['users']} # Create a dictionary for easy lookup
  for user in data['users']:
    friend_names = [user_dict[friend_id]['name'] for friend_id in user['friends']]
    print(f"User: {user['id']}.{user['name']} is friends with: {friend_names} and liked pages are: {user['liked_pages']}")
  print("pages information")
  for page in data['pages']:
    print(f"{page['id']}: {page['name']}")

display_users(data)

```
